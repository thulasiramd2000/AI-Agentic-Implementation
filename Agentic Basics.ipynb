```python
from openai import OpenAI
```

```python
openai=OpenAI()
```

```python
messages=[
    {"role": "user", "content": "Tell me a tech joke"},]
```

```python
response = openai.chat.completions.create(model="gpt-5.4-nano",messages=messages)
print(response.choices[0].message.content)
```

```python
question= "tell me a clear way to get placed in MAANG Companies. Respond only with one question"
messages=[
    {"role": "user", "content": question},]
```

```python
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
question=(response.choices[0].message.content)
print(question)
```

```python
messages=[
    {"role": "system", "content": question},]
messages
```

```python
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
answer=(response.choices[0].message.content)
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message= f"""
Here is a question: {question}

And here is the possible answer: {answer}

Please evaluate if the answer is correct or not. If it is correct, please provide a detailed explanation of why it is correct. If it is incorrect, please provide a detailed explanation of why it is incorrect and provide the correct answer.
"""
message
```

```python
messages= [{"role":"user", "content": message}]
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
print(response.choices[0].message.content)
```
